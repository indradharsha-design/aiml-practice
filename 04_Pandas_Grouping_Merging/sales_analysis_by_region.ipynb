```python
import pandas as pd
import numpy as np

data = {
    "OrderID": range(601, 613),
    "Region": [
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "North", "South", "East", "West"
    ],
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop",
        "Phone", "Tablet", "Laptop", "Phone",
        "Tablet", "Laptop", "Phone", "Tablet"
    ],
    "Salesperson": [
        "Asha", "Ben", "Chen", "Divya",
        "Asha", "Ben", "Chen", "Divya",
        "Asha", "Ben", "Chen", "Divya"
    ],
    "UnitsSold": [5, 12, 8, 6, 15, 10, 7, 11, 9, 4, 14, 13],
    "UnitPrice": [1200, 800, 400, 1200, 800, 400, 1200, 800, 400, 1200, 800, 400],
    "CustomerRating": [4.5, 4.0, 3.8, 4.2, 4.7, 4.1, 4.6, 3.9, 4.0, 4.3, 4.8, 4.4]
}

df = pd.DataFrame(data)
df["Revenue"] = df["UnitsSold"] * df["UnitPrice"]

df
```

```python
#Total revenue for the entire dataset
df["Revenue"].sum()
```

```python
#Total revenue by region
df.groupby("Region")["Revenue"].sum()
```

```python
#Total units sold by product
df.groupby("Product")["UnitsSold"].sum()
```

```python
#Average customer rating by region
df.groupby("Region")["CustomerRating"].mean().round(2)
```

```python
#Number of orders in each region
df.groupby("Region")["OrderID"].count()
```

```python
#Revenue by region and product
df.groupby(["Region", "Product"])["Revenue"].sum()
```

```python
#Each salesperson’s revenue and average rating
df.groupby("Salesperson").agg(
    TotalRevenue=("Revenue", "sum"),
    AverageRating=("CustomerRating", "mean")
).round(2)
```

```python
#Highest-revenue region
df.groupby("Region")["Revenue"].sum().idxmax()
```

```python
#Sort products by total revenue
df.groupby("Product")["Revenue"].sum().sort_values(
    ascending=False
)
```

```python
#Create the regional summary table
summary = df.groupby("Region").agg(
    TotalRevenue=("Revenue", "sum"),
    TotalUnitsSold=("UnitsSold", "sum"),
    AverageRating=("CustomerRating", "mean"),
    NumberOfOrders=("OrderID", "count")
).round(2)

summary
```

```python
#Filter regions with revenue above $15,000
summary[summary["TotalRevenue"] > 15000]
```

```python
#Reset the index
summary = summary.reset_index()
summary
```
