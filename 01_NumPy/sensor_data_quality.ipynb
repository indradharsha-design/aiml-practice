```python
import numpy as np

sensor_data = np.array([
    [24.5, 62.0],
    [26.1, 58.5],
    [99.9, 65.0],
    [23.8, -1.0],
    [25.4, 71.5],
    [np.nan, 60.0]
])

# Column 0 = Temperature (°C)
# Column 1 = Humidity (%)
```

A monitoring system records daily temperature and humidity readings from six locations.
Some readings are invalid or missing. Your job is to clean and analyze the data.

Task 1: Detect invalid readings
- Identify temperature readings above 50°C.
- Identify humidity readings below 0%.
- Identify missing values using np.isnan().

Task 2: Filter valid locations
Return rows where:
- Temperature is not missing and is at most 50°C.
- Humidity is between 0% and 100%, inclusive.

Task 3: Replace invalid readings
Create a copy of the dataset. Replace temperatures above 50°C with np.nan, and replace humidity values below 0% with np.nan.
Do not modify the original array.

Task 4: Calculate column averages
Calculate the mean temperature and humidity while ignoring missing values. Use np.nanmean().

Task 5: Conditional classification
Classify each temperature reading as:
- "Hot" if temperature is at least 26°C.
- "Normal" if temperature is below 26°C.
- "Missing" if the temperature is np.nan.
Constraint: Use NumPy operations and Boolean indexing. Avoid Python loops.

```python
# Temperature readings above 50°C
invalid_temp = sensor_data[:, 0] > 50
print("Invalid temperature:", invalid_temp)

# Humidity readings below 0%
invalid_humidity = sensor_data[:, 1] < 0
print("Invalid humidity:", invalid_humidity)

# Missing values
missing_values = np.isnan(sensor_data)
print("Missing values:\n", missing_values)
```

```python
valid_temp = (~np.isnan(sensor_data[:, 0])) & (sensor_data[:, 0] <= 50)
#~np.isnan(...) means the temperature is not missing.

valid_humidity = (
    (sensor_data[:, 1] >= 0) &
    (sensor_data[:, 1] <= 100)
)

valid_locations = sensor_data[valid_temp & valid_humidity]
print(valid_locations)
```

```python
clean_data = sensor_data.copy()

clean_data[clean_data[:, 0] > 50, 0] = np.nan
clean_data[clean_data[:, 1] < 0, 1] = np.nan

print(clean_data)
```

```python
#np.mean() normally propagates np.nan, making the result nan. 
# np.nanmean() ignores missing values when calculating the average.

mean_temperature = np.nanmean(clean_data[:, 0])
mean_humidity = np.nanmean(clean_data[:, 1])

print("Mean temperature:", mean_temperature)
print("Mean humidity:", mean_humidity)
```

```python
temperature = clean_data[:, 0]

classification = np.where(
    np.isnan(temperature),
    "Missing",
    np.where(temperature >= 26, "Hot", "Normal")
)

print(classification)

#np.where(condition, value_if_true, value_if_false) performs conditional selection without a Python loop. 
# The outer np.where() handles missing readings first, so they aren't incorrectly labelled "Normal"
```
