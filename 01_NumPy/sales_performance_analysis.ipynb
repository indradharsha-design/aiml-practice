```python
import numpy as np

sales = np.array([
    [120, 150, 135, 160],
    [95, 110, 105, 125],
    [180, 175, 190, 210],
    [140, 130, 155, 145],
    [105, 125, 115, 100]
])

targets = np.array([130, 140, 145, 150])
```

A company tracks quarterly sales across five regions.

Task 1: Quarterly performance
- Calculate total sales for each region.
- Calculate total sales for each quarter.
- Identify the quarter with the highest total sales.

Task 2: Target analysis
Each quarter has a sales target in targets.
- Compare the sales of every region against the corresponding quarterly target.
- Create a Boolean matrix indicating where sales meet or exceed the target.
- Count how many quarters each region met its target.

Task 3: Regional filtering
Identify regions whose total sales exceed the average total sales across all regions. Return their complete rows.

Task 4: Ranking
Calculate total sales for each region and sort the region totals from highest to lowest using NumPy, without Python loops.
Constraints: Use NumPy operations, Boolean indexing, broadcasting, sorting, and axis. No Python loops.

```python
total_region = sales.sum(axis=1)
total_quarter =sales.sum(axis=0)
highest_quarter = np.argmax(total_quarter)
print(total_region)
print(total_quarter)
print(highest_quarter)
```

```python
targets = np.array([130, 140, 145, 150])

# Compare each region's sales with each quarter's target
met_target = sales >= targets
print(met_target)

# Count quarters where each region met its target
quarters_met = met_target.sum(axis=1)
print(quarters_met)
```

```python
# Total sales for each region
total_region = sales.sum(axis=1)

# Average regional total
avg_region_sales = total_region.mean()

# Select regions above the average
above_avg = sales[total_region > avg_region_sales]

print("Average regional sales:", avg_region_sales)
print(above_avg)
```

```python
total_region = sales.sum(axis=1)

ranked_sales = np.sort(total_region)[::-1]

print(ranked_sales)

#np.sort(total_region) sorts the values in ascending order.
#[::-1] reverses that order to descending.
```

```python
#If you also want the region numbers in ranked order, use np.argsort():
ranked_regions = np.argsort(total_region)[::-1] + 1
print(ranked_regions)
```
