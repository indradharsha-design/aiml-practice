A subscription company tracks monthly usage for six customers.

Tasks
1. Calculate the average, minimum, and maximum for each of the three metrics.
2. Identify customers who logged in fewer than 10 days and raised more than 4 support tickets. Return their complete rows.
3. Create a Boolean mask identifying customers who spent at least $60.
4. Calculate a customer engagement score using this formula:
   \[
   \text{Score} = 2(\text{login days}) - 3(\text{support tickets})
   \]
5. Rank customers by engagement score from highest to lowest. Return both the customer numbers and their scores.
6. Normalize monthly spending to the range 0 to 1 using min-max normalization.
7. Calculate the correlation matrix between the three metrics using np.corrcoef().
Constraints: NumPy only. Use Boolean indexing, broadcasting or vectorized arithmetic, axis operations, sorting, normalization, and correlation. No Python loops.

```python
import numpy as np

# Columns: Login days, support tickets, monthly spending ($)
customers = np.array([
    [25, 1, 80],
    [8, 5, 45],
    [18, 2, 65],
    [5, 7, 30],
    [22, 1, 90],
    [12, 4, 50]
])

# Minimum expected spending for each customer
spending_target = 60
```

```python
averages = customers.mean(axis=0)
minimums = customers.min(axis=0)
maximums = customers.max(axis=0)

print("Averages:", averages)
print("Minimums:", minimums)
print("Maximums:", maximums)
```

```python
at_risk = customers[
    (customers[:, 0] < 10) &
    (customers[:, 1] > 4)
]

print(at_risk)
```

```python
spending_mask = customers[:, 2] >= spending_target
print(spending_mask)
print(customers[spending_mask])
```

```python
engagement_score = (
    2 * customers[:, 0] -
    3 * customers[:, 1]
)

print(engagement_score)
```

```python
ranking = np.argsort(engagement_score)[::-1]

print("Customer numbers:", ranking + 1)
print("Scores:", engagement_score[ranking])
```

```python
spending = customers[:, 2]

normalized_spending = (
    (spending - spending.min()) /
    (spending.max() - spending.min())
)

print(normalized_spending)
```

```python
correlation = np.corrcoef(customers, rowvar=False)
print(correlation)
```
