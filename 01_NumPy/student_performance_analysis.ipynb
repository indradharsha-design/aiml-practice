```python
import numpy as np

scores = np.array([
    [78, 85, 92, 67],
    [88, 76, 95, 81],
    [69, 90, 84, 73],
    [94, 82, 79, 87],
    [56, 68, 72, 61],
    [83, 89, 91, 85]
])

subject_bonus = np.array([2, 0, 3, 1])
```

A school wants to analyze assessment scores across four subjects for six students.
Tasks
1. Calculate the average score for each student.
2. Identify students whose average score is above the class average. Use Boolean indexing.
3. Find the subject with the lowest average score.
4. Apply subject_bonus to every student's scores using broadcasting. Calculate the adjusted score matrix.
5. Reshape the adjusted scores into a (3 * 8) matrix and calculate the mean of each row.
6. Identify students who scored at least 90 in any subject. Return their original score rows using Boolean indexing.
Constraints: Use NumPy operations only. No Python loops or manual calculations.

```python
student_avg= scores.mean(axis=1)
print(student_avg)
```

```python
class_avg = scores.mean()
best_student_avg = student_avg[student_avg > class_avg]
print(best_student_avg)
```

```python
sub_avg= scores.mean(axis=0)
lowest_subject = np.argmin(sub_avg)
print(lowest_subject)
```

```python
subject_bonus = np.array([2, 0, 3, 1])
adjusted_scores= scores+subject_bonus
print(adjusted_scores)
```

```python
reshaped_scores = adjusted_scores.reshape(3, 8)
print(reshaped_scores)

row_avg = reshaped_scores.mean(axis=1)
print("Row averages:", row_avg)
```

```python
qualified = np.any(scores >= 90, axis=1)
print(qualified)
qualified_students = scores[qualified]
print(qualified_students)


#scores[scores >= 90] returns matching individual values.
#scores[np.any(scores >= 90, axis=1)] returns complete rows containing at least one matching value.
```
