A company measures three quality indicators for products from four factories.
Each row represents a factory. Each column represents a quality indicator.
Tasks
1. Calculate the average score for each quality indicator.
2. Calculate a weighted quality score for each factory. Multiply each indicator by its corresponding weight, then sum across each row.
3. Identify factories with a weighted score of at least 85. Return their complete rows.
4. Calculate each factory's difference from the target for all three indicators using broadcasting.
5. Find the highest indicator score for each factory and identify its column index.
6. Rank factories from highest to lowest weighted quality score.

```python
import numpy as np

quality = np.array([
    [82, 75, 91],
    [88, 84, 79],
    [72, 90, 85],
    [95, 87, 93]
])

weights = np.array([0.4, 0.35, 0.25])
targets = np.array([80, 80, 80])
```

```python
indicator_avg = quality.mean(axis=0)
print(indicator_avg)
```

```python
weighted_scores = (quality * weights).sum(axis=1)
print(weighted_scores)
```

```python
selected = quality[weighted_scores >= 85]
print(selected)
```

```python
difference = quality - targets
print(difference)
```

```python
highest_scores = quality.max(axis=1)
highest_indices = quality.argmax(axis=1)

print(highest_scores)
print(highest_indices)
```

```python
ranking = np.argsort(weighted_scores)[::-1]
print(ranking + 1)
print(weighted_scores[ranking])
```
