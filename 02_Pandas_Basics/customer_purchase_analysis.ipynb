Task 4: Handle missing data
12. Make a copy of the DataFrame and set Customer 103's satisfaction to NaN.
13. Count missing values in each column.
14. Fill missing satisfaction scores with the mean satisfaction score.

```python
df_missing = df.copy()

df_missing.loc[
    df_missing["CustomerID"] == 103,
    "Satisfaction"
] = np.nan
```

```python
df_missing.isna().sum()
```

```python
mean_satisfaction = df_missing["Satisfaction"].mean()

df_missing["Satisfaction"] = (
    df_missing["Satisfaction"].fillna(mean_satisfaction)
)

print("Mean satisfaction:", mean_satisfaction)
print(df_missing[["CustomerID", "Satisfaction"]])
```
