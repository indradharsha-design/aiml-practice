```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "EmployeeID": [201, 202, 203, 204, 205, 206, 207, 208],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "HR"
    ],
    "Experience": [5, 2, 8, 4, 7, 3, 6, 1],
    "Salary": [95000, 58000, 125000, 82000, 76000, 65000, 110000, 48000],
    "PerformanceScore": [4.5, 3.2, 4.8, 3.9, 4.1, 3.5, 4.6, 2.9],
    "ProjectsCompleted": [12, 5, 18, 9, 11, 7, 15, 3]
})

df
```

Task 1: Filtering
1. Find employees earning more than $80,000.
2. Find IT employees with at least 5 years of experience.
3. Find employees with a performance score of 4.0 or higher and at least 10 completed projects.
4. Find employees whose department is either HR or Finance.

```python
df[df["Salary"] > 80000]
```

```python
df[(df["Department"]=="IT") & (df["Experience"]>=5)]
```

```python
df[(df["PerformanceScore"]>=4.0) & (df["ProjectsCompleted"]>=10)]
```

```python
#df[(df["Department"]=="HR") | (df["Department"]=="Finance")]
df[df["Department"].isin(["HR", "Finance"])]
```

Task 2: Conditional columns
5. Create BonusEligible, set to True when PerformanceScore >= 4.0.
6. Create ExperienceLevel:
- Junior: fewer than 3 years
- Mid-Level: 3–6 years
- Senior: more than 6 years
7. Create SalaryBand: "High" for salary at least $100,000, otherwise "Standard".

```python
df["BonusEligible"] = df["PerformanceScore"] >= 4.0
df[["EmployeeID", "Salary", "BonusEligible"]]
```

```python
df["ExperienceLevel"] = np.where(
    df["Experience"] < 3, "Junior",
    np.where(df["Experience"] <= 6, "Mid-Level", "Senior")
)

print(df[["EmployeeID", "Experience", "ExperienceLevel"]])
```

```python
df["SalaryBand"] = np.where(
    df["Salary"] >= 100000, "High", "Standard"
)
print(df[["EmployeeID", "Salary", "SalaryBand"]])
```

8. Sort employees by salary, highest first.
